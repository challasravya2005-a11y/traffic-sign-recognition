```python
# TRAFFIC SIGN RECOGNITION - COMPLETE PROJECT

!pip -q install torch torchvision

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
import numpy as np

print("Starting Traffic Sign Recognition Project...")

# 1. Image preprocessing
transform = transforms.Compose([
    transforms.Resize((32, 32)),
    transforms.ToTensor()
])

# 2. Download real traffic sign dataset
train_data = datasets.GTSRB(
    root="./data",
    split="train",
    download=True,
    transform=transform
)

test_data = datasets.GTSRB(
    root="./data",
    split="test",
    download=True,
    transform=transform
)

# Use a smaller portion to finish faster
train_data = Subset(train_data, range(5000))
test_data = Subset(test_data, range(1000))

train_loader = DataLoader(train_data, batch_size=64, shuffle=True)
test_loader = DataLoader(test_data, batch_size=64, shuffle=False)

print("Dataset loaded successfully!")
print("Training images:", len(train_data))
print("Testing images:", len(test_data))

# 3. CNN model
class TrafficSignCNN(nn.Module):
    def __init__(self):
        super().__init__()

        self.network = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Conv2d(32, 64, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),

            nn.Flatten(),
            nn.Linear(64 * 8 * 8, 128),
            nn.ReLU(),
            nn.Linear(128, 43)
        )

    def forward(self, x):
        return self.network(x)

model = TrafficSignCNN()

# 4. Training setup
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

print("Training started...")

# 5. Train
for epoch in range(3):
    model.train()
    correct = 0
    total = 0

    for images, labels in train_loader:
        optimizer.zero_grad()

        output = model(images)
        loss = criterion(output, labels)

        loss.backward()
        optimizer.step()

        _, predicted = torch.max(output, 1)
        total += labels.size(0)
        correct += (predicted == labels).sum().item()

    accuracy = 100 * correct / total
    print(f"Epoch {epoch+1}/3 - Training Accuracy: {accuracy:.2f}%")

# 6. Test
model.eval()
correct = 0
total = 0

with torch.no_grad():
    for images, labels in test_loader:
        output = model(images)
        _, predicted = torch.max(output, 1)

        total += labels.size(0)
        correct += (predicted == labels).sum().item()

test_accuracy = 100 * correct / total

print("\n==============================")
print("TRAFFIC SIGN RECOGNITION RESULT")
print("==============================")
print(f"Test Accuracy: {test_accuracy:.2f}%")

# 7. Show sample predictions
images, labels = next(iter(test_loader))

with torch.no_grad():
    output = model(images)
    _, predictions = torch.max(output, 1)

plt.figure(figsize=(10, 7))

for i in range(9):
    plt.subplot(3, 3, i + 1)
    image = images[i].permute(1, 2, 0).numpy()
    plt.imshow(image)
    plt.title(
        f"Actual: {labels[i].item()}\n"
        f"Predicted: {predictions[i].item()}"
    )
    plt.axis("off")

plt.tight_layout()
plt.show()

print("\nProject completed successfully!")
```
